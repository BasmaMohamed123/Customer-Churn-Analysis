## Task 1: Data Cleaning and Preprocessing

**Project:** Saiket Systems - Data Analysis Internship

### Objectives
- Inspect the dataset structure.
- Identify missing values and duplicates.
- Handle missing and inconsistent data.
- Convert categorical variables into numerical representations.
- Prepare the dataset for further analysis.

**Dataset:** Telco Customer Churn Dataset


```python

# Import Libraries
import pandas as pd
import numpy as np

# Load Dataset
df = pd.read_csv(
    r"D:\Saiket_Data_Analysis\data\Telco_Customer_Churn_Dataset .csv"
)

# Display First 5 Rows
print("First 5 Rows:")
display(df.head())

# Dataset Dimensions
print("\nDataset Shape:", df.shape)

# Dataset Information
print("\nDataset Information:")
df.info()

# Statistical Summary
print("\nStatistical Summary:")
display(df.describe(include="all").T)
```

```python
# Data Quality Checks

print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nUnique Values:")
print(df.nunique())

print("\nTotalCharges values before conversion:")
print(df["TotalCharges"].tail(15).to_list())
```

```python
# Clean TotalCharges column

df["TotalCharges"] = df["TotalCharges"].str.strip()

df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

print("Missing Values After Conversion:")
print(df["TotalCharges"].isnull().sum())

print("\nRows with Missing TotalCharges:")
display(df[df["TotalCharges"].isnull()])
```

```python
""# Handle missing TotalCharges values

df["TotalCharges"] = df["TotalCharges"].fillna(0)

print("Missing Values After Cleaning:")
print(df.isnull().sum().sum())

print("\nTotalCharges Data Type:")
print(df["TotalCharges"].dtype)

print("\nUpdated Dataset Shape:")
print(df.shape)
```

```python
# Check categorical and numerical columns

categorical_columns = df.select_dtypes(include="str").columns

numerical_columns = df.select_dtypes(include="number").columns

print("Categorical Columns:")
print(list(categorical_columns))

print("\nNumerical Columns:")
print(list(numerical_columns))

print("\nCategorical Values:")
for col in categorical_columns:
    print(f"\n{col}: {df[col].unique()}")

print("\nNumerical Summary:")
display(df[numerical_columns].describe().T)
```

```python
# Prepare data for encoding

df_encoded = df.drop(columns=["customerID"]).copy()

# Encode binary categorical columns
binary_columns = [
    "gender",
    "Partner",
    "Dependents",
    "PhoneService",
    "PaperlessBilling",
    "Churn"
]

for col in binary_columns:
    df_encoded[col] = df_encoded[col].map({
        "Yes": 1,
        "No": 0,
        "Male": 1,
        "Female": 0
    })

# One-hot encode multi-category columns
df_encoded = pd.get_dummies(
    df_encoded,
    columns=[
        "MultipleLines",
        "InternetService",
        "OnlineSecurity",
        "OnlineBackup",
        "DeviceProtection",
        "TechSupport",
        "StreamingTV",
        "StreamingMovies",
        "Contract",
        "PaymentMethod"
    ],
    dtype=int
)

print("Original Dataset Shape:", df.shape)
print("Encoded Dataset Shape:", df_encoded.shape)

print("\nEncoded Dataset Preview:")
display(df_encoded.head())

print("\nData Types After Encoding:")
print(df_encoded.dtypes.value_counts())
```

```python
# Final Data Cleaning Validation

print("Dataset Shape:", df_encoded.shape)

print("\nMissing Values:", df_encoded.isnull().sum().sum())

print("\nDuplicate Rows:", df_encoded.duplicated().sum())

print("\nNon-Numeric Columns:")
print(df_encoded.select_dtypes(exclude="number").columns.tolist())

print("\nTarget Distribution:")
print(df_encoded["Churn"].value_counts())

print("\nFinal Dataset Preview:")
display(df_encoded.head())
```

```python
print("Original Dataset Duplicates:", df.duplicated().sum())
print("Encoded Dataset Duplicates:", df_encoded.duplicated().sum())

print("\nOriginal Dataset Shape:", df.shape)
print("Encoded Dataset Shape:", df_encoded.shape)
```

```python
# Save cleaned datasets

df.to_csv(
    r"D:\Saiket_Data_Analysis\data\Telco_Customer_Churn_Cleaned.csv",
    index=False
)

df_encoded.to_csv(
    r"D:\Saiket_Data_Analysis\data\Telco_Customer_Churn_Encoded.csv",
    index=False
)

print("Both datasets saved successfully!")
```
